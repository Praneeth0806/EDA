```python
# Step 1 : Calculate Q1
q1 = np.quantile(pwage,0.25)
# Step 2 : Calculate Q2
q2 = np.quantile(pwage,0.5)
# Step 3 : Calculate Q3
q3 = np.quantile(pwage,0.75)
# Step 4 : Calculate iqr = q3-q1
iqr = q3-q1
# Step 5 : Calculate lb = q1-1.5*iqr
lower_bound = q1-(1.5*iqr)
# Step 6 : Calculate ub = q3+1.5*iqr
upper_bound = q3+(1.5*iqr)
# Step 7 : Calculate con1
condition_1 = pwage<lower_bound
# Step 8 : Calculate con2
condition_2 = pwage>upper_bound
# Step 9 : Calculate con
final_condition = condition_1 | condition_2
outliers_df = visa[final_condition]
# Step 10 : Calculate count
len(outliers_df)
```

```python
outliers_list = outliers_df['prevailing_wage'].values.tolist()
outliers_list
```

**outlier filling using method median**

```python
wage_median = visa['prevailing_wage'].median()
wage_data = visa['prevailing_wage']
new_wage_data = []
for i in wage_data:
    if i in outliers_list:
        new_wage_data.append(wage_median)
    else:
        new_wage_data.append(i)
```

**another approach winsorization**

- whatever the outliers less than lb fill with lb

- whatever the outliers greater than ub fill with ub

- in numpy we have a method called **clip**

```python
# 1) get wage data values
# 2) calculate lb
# 3) calculate ub
# 4) check the parameters and apply in numpy np.clip()
# 5) again implement above visualization analysis
```

```python
pre_wage = visa['prevailing_wage']
# Step 1 : Calculate Q1
q1 = np.quantile(pre_wage,0.25)
# Step 2 : Calculate Q2
q2 = np.quantile(pre_wage,0.5)
# Step 3 : Calculate Q3
q3 = np.quantile(pre_wage,0.75)
# Step 4 : Calculate iqr = q3-q1
iqr = q3-q1
# Step 5 : Calculate lb = q1-1.5*iqr
lower_bound = q1-(1.5*iqr)
# Step 6 : Calculate ub = q3+1.5*iqr
upper_bound = q3+(1.5*iqr)
# check the parameters and apply in numpy np.clip()
wage_data_clip = np.clip(pre_wage,a_min=lower_bound,a_max=upper_bound)
```

```python
plt.figure(figsize=(10,6))
plt.subplot(2,2,1).hist(pre_wage)
plt.subplot(2,2,2).boxplot(pre_wage,vert=False)
plt.subplot(2,2,3).hist(wage_data_clip)
plt.subplot(2,2,4).boxplot(wage_data_clip,vert=False)
plt.show()
```

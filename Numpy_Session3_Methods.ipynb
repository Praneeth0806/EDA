**np.random.randint**

```python
import numpy as np
np.random.randint(1,10)

# np.random.randint(low, high=None, size=None, dtype=int)
# np.random.randint(1,10)
```

```python
np.random.randint(1,100)
# size= 10
# size= (2,2)
```

```python
np.random.randint(1,100,10)
#np.random.randint(low,high,size)
# low=1  high=100 size=10
```

```python
shape=(3,2,2)
np.random.randint(1,100,shape)
##np.random.randint(low,high,size,dtype)
```

**np.random.random**

```python
random.random()
```

```python
random.randint ====> np.random.randint
random.random  ====> np.random.random
```

```python
np.random.random(size=10)
```

```python
np.random.random(size=(2,2))
# 0 to 1
# [0,1)
```

```python
np.random.rand(10)
```

```python
np.random.rand(2,2)
# 0 to 1
```

```python
np.random.randn(5)  # standard normal
```

**choice**

```python
random.randint ==== np.random.randint
random.random  ==== np.random.random
random.choice  ==== np.random.choice
```

```python
l=list(range(1,10))
random.choice(l)
```

```python
np.random.choice(l)
```

```python
np.random.choice(l,size=(2,2))
#np.random.choice(a, size=None)
```

- always focus argumnets order

- size means always provide a shape in a tuple

- always read the doc string to understand the method 

**np.random.normal**

```python
data=np.random.normal(100,2,200)
```

```python
np.mean(data),np.std(data)
```

```python
import matplotlib.pyplot as plt
plt.hist(data)
```

**np.random.exponetinatl**

```python
data=np.random.exponential(2,500)
```

```python
plt.hist(data)
plt.show()
```

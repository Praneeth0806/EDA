```python
SUM = np.sum(arr1)
count = np.count_nonzero(arr1)
mean = SUM/count
```

```python
# step 1 : arr1 (xi means array)
print(arr1)
# step 2 : mean (xbar means mean of array)
mean = np.mean(arr1)
print(mean)
# step 3 : arr1-mean (calculate xi-xbar in the form of array)
arr2 = arr1-mean
print(arr2)
# step 4 : square ( square of xi - xbar)
arr3 = np.square(arr2)
print(arr3)
# step 5 : sum ( numerator = sum of square of xi - xbar)
arr4 = np.sum(arr3)
print(arr4)
# step 6 : calculation (numerator / count)
var = arr4 / np.count_nonzero(arr1)
print(var)
```

```python
a = np.sum(np.square(arr1-np.mean(arr1))) / np.count_nonzero(arr1)
a
```

**standard deviation**

```python
np.std(arr1)
```

```python
# step 1 : arr1 (xi means array)
print(arr1)
# step 2 : mean (xbar means mean of array)
mean = np.mean(arr1)
print(mean)
# step 3 : arr1-mean (calculate xi-xbar in the form of array)
arr2 = arr1-mean
print(arr2)
# step 4 : square ( square of xi - xbar)
arr3 = np.square(arr2)
print(arr3)
# step 5 : sum ( numerator = sum of square of xi - xbar)
arr4 = np.sum(arr3)
print(arr4)
# step 6 : calculation (numerator / count)
var = arr4 / np.count_nonzero(arr1)
print(var)
# step 7 : square root of var for standard deviation
std = np.sqrt(var)
print(std)
```

```python
np.sqrt((np.sum(np.square(arr1-np.mean(arr1))))/np.count_nonzero(arr1))
```

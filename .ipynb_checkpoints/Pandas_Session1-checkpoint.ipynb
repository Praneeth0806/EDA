**inplace**

- when we do the operation the result output is not overwritten

- means the original dataframe is as it is

- in order to do this we need to use **inplace**

- by default inplace = False

- make inplace = True if you want overwrite the result

```python
df
```

```python
df.drop('Marks',axis=1,inplace=True)
```

```python
df
```

```python
df.drop(1,axis=0,inplace=True)
```

```python
df
```

```python
# create one more df dataframe
# x= 1 to 10
# y = square of x
# z = cube of x
```

```python
x = list(range(1,11))
print(x)
y = np.square(x)
print(y)
z = x*y
print(z)
```

```python
column = ['x','y','z']
df_11 = pd.DataFrame(zip(x,y,z),columns=column)
df_11
```

```python
import numpy as np
df1 = pd.DataFrame()
df1['x']=range(1,11)
df1['y'] = np.square(range(1,11))
df1['z'] = df1['x'] * df1['y']
df1
```

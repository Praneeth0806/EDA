```python
# Step 1 : Create a folder
folder_name = 'Frequency'
os.makedirs(folder_name,exist_ok=True)

# step 2 : make a new path with the created folder
directory = os.getcwd()
new_dir = os.path.join(directory,folder_name)

for i in cat[1:]:

    # step 3: Create a dataframe------------------------------------------------------
    keys_11 = visa_df[i].value_counts().keys()
    values_11 = visa_df[i].value_counts().values
    df = pd.DataFrame(zip(keys_11,values_11),columns=['Class','Frequency'])
    
    # step 4 : make the new path ----------------------------------------------------------
    file_name = f'{i}_df.csv'
    path = os.path.join(new_dir,file_name)
    
    # Step 5: save into dataframe
    df.to_csv(path,index=False)
```

**bar chart**

```python
# Step 1 : Create a folder
# step 2 : make a new path with the created folder
# step 3: Create a dataframe
# step 4 : make the new path
# Step 5: save into dataframe
```

```python
# Step 1 : Create a folder
folder = 'bar_chart'
os.makedirs(folder,exist_ok=True)
# step 2 : make a new path with the created folder
current_dir = os.getcwd()
new_dir = os.path.join(current_dir,folder)
for i in cat[1:]:
    # step 3: Create a dataframe
    plt.figure(figsize=(10,4))
    key = visa_df[i].value_counts().keys()
    value = visa_df[i].value_counts().values
    plt.bar(key,value)
    plt.title(f'{i} barchart')
    # step 4 : make the new path
    file = f'{i}_barchart.png'
    new_path = os.path.join(new_dir,file)
    # Step 5: save into dataframe
    plt.savefig(new_path)
    plt.close()
```

**pie chart**

```python
# Step 1 : Create a folder
folder = 'pie_chart'
os.makedirs(folder,exist_ok=True)
# step 2 : make a new path with the created folder
current_dir = os.getcwd()
new_dir = os.path.join(current_dir,folder)
for i in cat[1:]:
    # step 3: Create a dataframe
    plt.figure(figsize=(10,4))
    key = visa_df[i].value_counts().keys()
    value = visa_df[i].value_counts().values
    plt.pie(value,labels=key,autopct='%1.2f%%',startangle=90)
    plt.title(f'{i} piechart')
    # step 4 : make the new path
    file = f'{i}_piechart.png'
    new_path = os.path.join(new_dir,file)
    # Step 5: save into dataframe
    plt.savefig(new_path)
    plt.close()
```
